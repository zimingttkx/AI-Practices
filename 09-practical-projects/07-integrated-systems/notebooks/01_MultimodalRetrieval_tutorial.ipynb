## 2. 多模态文档

多模态文档可以同时包含文本和图像信息。

```python
from multimodal_retriever import MultimodalDocument, ModalityType

# 创建纯文本文档
text_doc = MultimodalDocument(content="机器学习是人工智能的核心技术")
print(f"文本文档: {text_doc}")
print(f"模态类型: {text_doc.modality}")
print(f"有文本: {text_doc.has_text}, 有图像: {text_doc.has_image}")
```

```python
# 创建图像文档
image_data = np.random.rand(64, 64, 3).astype(np.float32)
image_doc = MultimodalDocument(image_data=image_data, content="一张示例图片")

print(f"图像文档: {image_doc}")
print(f"模态类型: {image_doc.modality}")
print(f"图像形状: {image_doc.image_data.shape}")
```

## 3. 编码器

编码器将文本和图像转换为向量表示。

```python
from multimodal_retriever import TextEncoder, ImageEncoder, CLIPEncoder

# 文本编码器
text_encoder = TextEncoder(dim=128)
text_emb = text_encoder.encode("深度学习使用神经网络")
print(f"文本嵌入形状: {text_emb.shape}")
print(f"向量范数: {np.linalg.norm(text_emb[0]):.4f} (已归一化)")
```

```python
# 图像编码器
image_encoder = ImageEncoder(dim=128)
image = np.random.rand(32, 32, 3).astype(np.float32)
image_emb = image_encoder.encode(image)
print(f"图像嵌入形状: {image_emb.shape}")
```

```python
# CLIP联合编码器 - 将文本和图像映射到同一空间
clip_encoder = CLIPEncoder(dim=256)

text_vec = clip_encoder.encode_text("一只猫")
image_vec = clip_encoder.encode_image(np.random.rand(64, 64, 3))

# 计算相似度
similarity = np.dot(text_vec[0], image_vec[0])
print(f"文本-图像相似度: {similarity:.4f}")
```

## 4. 多模态检索器

```python
from multimodal_retriever import MultimodalRetriever

# 创建检索器
retriever = MultimodalRetriever()

# 添加文档
docs = [
    MultimodalDocument(content="Python是一种流行的编程语言"),
    MultimodalDocument(content="机器学习需要大量训练数据"),
    MultimodalDocument(content="深度学习是机器学习的子集"),
    MultimodalDocument(content="神经网络模拟人脑结构"),
    MultimodalDocument(content="自然语言处理研究文本理解"),
]

retriever.add_documents(docs)
print(f"文档数量: {retriever.num_documents}")
```

```python
# 文本检索
results = retriever.search_by_text("神经网络和深度学习", top_k=3)

print("检索结果:")
for i, r in enumerate(results, 1):
    print(f"  {i}. [{r.score:.4f}] {r.document.content}")
```

## 5. 图像检索

```python
# 添加带图像的文档
retriever.clear()

# 模拟不同类型的图像
np.random.seed(42)
cat_img = np.random.rand(64, 64, 3).astype(np.float32)
dog_img = np.random.rand(64, 64, 3).astype(np.float32) + 0.1
car_img = np.random.rand(64, 64, 3).astype(np.float32) + 0.2

retriever.add_document(MultimodalDocument(content="可爱的猫咪", image_data=cat_img))
retriever.add_document(MultimodalDocument(content="活泼的小狗", image_data=dog_img))
retriever.add_document(MultimodalDocument(content="红色跑车", image_data=car_img))

print(f"添加了 {retriever.num_documents} 个图像文档")
```

```python
# 用文本检索图像
results = retriever.search_by_text("宠物", top_k=2)

print("文本→图像检索:")
for r in results:
    print(f"  [{r.score:.4f}] {r.document.content}")
```

```python
# 用图像检索相似图像
query_img = cat_img + np.random.rand(64, 64, 3) * 0.05  # 添加噪声
results = retriever.search_by_image(query_img, top_k=2)

print("图像→图像检索:")
for r in results:
    print(f"  [{r.score:.4f}] {r.document.content}")
```
