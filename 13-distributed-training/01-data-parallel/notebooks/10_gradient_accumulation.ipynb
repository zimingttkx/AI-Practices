# Gradient Accumulation Tutorial

## Overview

Simulate larger batch sizes without increasing memory usage.

### Formula
$$\text{Effective Batch} = \text{Micro Batch} \times \text{Accumulation Steps} \times \text{Num GPUs}$$

```python
import torch

def train_with_accumulation(model, dataloader, optimizer, accumulation_steps=4):
    """Training loop with gradient accumulation."""
    model.train()
    optimizer.zero_grad()
    
    for i, (data, target) in enumerate(dataloader):
        # Forward pass
        output = model(data)
        loss = torch.nn.functional.cross_entropy(output, target)
        
        # Scale loss by accumulation steps
        loss = loss / accumulation_steps
        loss.backward()
        
        # Update weights every accumulation_steps
        if (i + 1) % accumulation_steps == 0:
            optimizer.step()
            optimizer.zero_grad()
            print(f"Step {i+1}: Updated weights")
```

## Summary

| Micro Batch | Accum Steps | GPUs | Effective Batch |
|-------------|-------------|------|----------------|
| 8 | 4 | 8 | 256 |
| 4 | 8 | 16 | 512 |
| 2 | 16 | 32 | 1024 |
