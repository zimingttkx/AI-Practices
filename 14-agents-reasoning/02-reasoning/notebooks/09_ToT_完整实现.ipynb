## 1. TreeOfThoughts 类

### 1.1 创建实例

```python
# 默认配置（BFS）
tot_default = TreeOfThoughts()
print("默认 ToT 已创建")

# 自定义配置
tot_beam = TreeOfThoughts(
    search_strategy=BeamSearch(beam_width=3),
    generator=SimpleThoughtGenerator(),
    evaluator=SimpleThoughtEvaluator()
)
print("Beam Search ToT 已创建")
```

### 1.2 解决问题

```python
# 使用 ToT 解决问题
problem = "使用4,5,6,7四个数字，通过加减乘除得到24"
result = tot_default.solve(problem)

print(f"问题: {problem}")
print(f"找到解: {result.solution_node is not None}")
print(f"探索节点数: {result.total_nodes}")
```
