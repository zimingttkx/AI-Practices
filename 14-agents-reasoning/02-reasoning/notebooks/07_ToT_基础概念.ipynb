## 1. 什么是 ToT

### 1.1 核心思想

Tree of Thoughts 将推理过程建模为树搜索，探索多条推理路径。

```python
print("""
Tree of Thoughts 核心概念：

                    [问题]
                       │
           ┌───────────┼───────────┐
           │           │           │
        [思路A]     [思路B]     [思路C]
           │           │           │
       ┌───┴───┐   ┌───┴───┐   ┌───┴───┐
       │       │   │       │   │       │
     [A1]    [A2] [B1]   [B2] [C1]   [C2]
       │       │   │       │   │       │
      ...     ... ...     ... ...     ...
                       │
                   [最优解]

关键特点：
  1. 探索多条推理路径
  2. 评估每个节点的价值
  3. 可以回溯和剪枝
  4. 找到最优解决方案
""")
```

### 1.2 适用场景

```python
print("""
ToT 适用场景：

✓ 创意写作（多种开头/结局）
✓ 数学证明（多种证明路径）
✓ 游戏策略（多步决策）
✓ 规划问题（多种方案）
✓ 代码生成（多种实现）

不适用场景：
✗ 简单问答
✗ 事实查询
✗ 单步推理
""")
```

## 2. ToT vs CoT

### 2.1 对比分析

```python
print("""
┌──────────────┬─────────────────┬─────────────────┐
│     特性     │      CoT        │      ToT        │
├──────────────┼─────────────────┼─────────────────┤
│ 推理结构     │ 线性链          │ 树形结构        │
│ 路径数量     │ 单条            │ 多条            │
│ 回溯能力     │ 无              │ 有              │
│ 计算成本     │ 低              │ 高              │
│ 适用问题     │ 简单推理        │ 复杂探索        │
│ 可解释性     │ 高              │ 更高            │
└──────────────┴─────────────────┴─────────────────┘

CoT: 问题 → 步骤1 → 步骤2 → 步骤3 → 答案

ToT: 问题 → [多个思路] → [评估] → [选择] → [展开] → 最优答案
""")
```

## 3. 核心数据结构

### 3.1 NodeStatus 节点状态

```python
print("节点状态：")
for status in NodeStatus:
    print(f"  {status.name}: {status.value}")
```

### 3.2 ThoughtNode 思维节点

```python
# 创建根节点
root = ThoughtNode(
    thought="解决24点游戏：使用4,5,6,7得到24",
    depth=0
)

print(f"根节点：")
print(f"  内容: {root.thought}")
print(f"  深度: {root.depth}")
print(f"  状态: {root.status}")
```

```python
# 创建子节点
child1 = ThoughtNode(
    thought="尝试 (4+5) = 9",
    depth=1,
    parent=root,
    value=0.6
)

child2 = ThoughtNode(
    thought="尝试 (6-4) = 2",
    depth=1,
    parent=root,
    value=0.7
)

root.children = [child1, child2]

print(f"子节点：")
for i, child in enumerate(root.children, 1):
    print(f"  {i}. {child.thought} (分数: {child.value})")
```

```python
# 节点操作
print(f"根节点是否为叶子: {root.is_leaf}")
print(f"子节点1是否为叶子: {child1.is_leaf}")
print(f"根节点子节点数: {len(root.children)}")
```
