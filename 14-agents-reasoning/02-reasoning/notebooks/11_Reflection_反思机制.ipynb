## 2. 核心组件

### 2.1 EvaluationResult 评估结果

```python
print("评估结果类型：")
for result in EvaluationResult:
    print(f"  {result.name}: {result.value}")
```

### 2.2 ReflectionStep 反思步骤

```python
# 创建反思步骤
step1 = ReflectionStep(
    iteration=1,
    response="机器学习是一种技术",
    evaluation=EvaluationResult.PARTIALLY_CORRECT,
    feedback="回答太简短，缺少细节",
    score=0.4,
    improvements=["添加定义", "举例说明"]
)

print(f"反思步骤 1：")
print(f"  响应: {step1.response}")
print(f"  评估: {step1.evaluation.value}")
print(f"  分数: {step1.score}")
print(f"  改进: {step1.improvements}")
```

### 2.3 ReflectionResult 最终结果

```python
# 创建最终结果
result = ReflectionResult(
    question="什么是机器学习？",
    initial_response="机器学习是一种技术",
    final_response="机器学习是AI的分支，让计算机从数据中学习模式",
    steps=[step1],
    total_iterations=2,
    improved=True
)

print(f"反思结果：")
print(f"  初始: {result.initial_response}")
print(f"  最终: {result.final_response}")
print(f"  改进: {result.improved}")
```

## 3. 自我评估

### 3.1 SimpleSelfEvaluator

```python
evaluator = SimpleSelfEvaluator()

# 评估响应
result, feedback, score = evaluator.evaluate(
    question="什么是深度学习？",
    response="深度学习是机器学习的子领域"
)

print(f"评估结果：")
print(f"  结果: {result.value}")
print(f"  反馈: {feedback}")
print(f"  分数: {score}")
```

## 4. 错误检测与纠正

### 4.1 ErrorDetector

```python
detector = ErrorDetector()

# 测试不同响应
responses = [
    "这个答案是正确的。",
    "我不确定这是否正确。",
    "这可能是错误的。",
]

print("错误检测：")
for resp in responses:
    errors = detector.detect(resp)
    print(f"  '{resp[:25]}...'")
    print(f"    问题: {errors if errors else '无'}")
```

### 4.2 CorrectionStrategy

```python
corrector = SimpleCorrectionStrategy()

# 生成纠正建议
corrected = corrector.correct(
    question="问题",
    response="答案是42",
    feedback="缺少推理过程"
)

print(f"纠正结果: {corrected}")
```

## 5. Reflection 类

```python
# 创建 Reflection 实例
reflection = Reflection(
    max_iterations=3,
    target_score=0.8
)

print(f"Reflection 配置：")
print(f"  最大迭代: {reflection._max_iterations}")
print(f"  目标分数: {reflection._target_score}")
```

```python
# 运行反思
result = reflection.reflect(
    question="解释神经网络",
    initial_response="神经网络是计算模型"
)

print(f"反思结果：")
print(f"  初始: {result.initial_response}")
print(f"  最终: {result.final_response}")
print(f"  迭代: {result.total_iterations}")
```
