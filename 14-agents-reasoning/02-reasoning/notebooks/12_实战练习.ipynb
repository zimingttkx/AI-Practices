### 任务：创建工具并模拟 ReAct 执行

```python
# 示例：创建工具
def search_func(query):
    data = {
        "python": "Python由Guido van Rossum于1991年创建",
        "java": "Java由Sun公司于1995年发布",
    }
    for k, v in data.items():
        if k in query.lower():
            return v
    return "未找到"

search_tool = SimpleTool("search", "搜索信息", search_func)
print(f"测试: {search_tool.execute('Python创建者')}")
```

## 练习3: ToT 创意问题

### 任务：使用 ToT 解决24点游戏

```python
# 示例：创建思维树
root = ThoughtNode(thought="使用2,3,4,6得到24", depth=0)

# 第一层：尝试不同运算
child1 = ThoughtNode(thought="2×3=6", depth=1, parent=root, value=0.5)
child2 = ThoughtNode(thought="4×6=24", depth=1, parent=root, value=0.9)
child3 = ThoughtNode(thought="2+3=5", depth=1, parent=root, value=0.3)

root.children = [child1, child2, child3]

print("思维树：")
print(f"根: {root.thought}")
for c in root.children:
    print(f"  └─ {c.thought} (分数:{c.value})")
```

## 练习4: 策略组合

### 任务：实现 CoT + Self-Consistency

```python
# 示例：模拟多次采样
question = "一个班有40人，男生比女生多8人，男生有多少人？"

# 模拟5次采样结果
paths = [
    SampledPath("设女生x，男生x+8，2x+8=40，x=16，男生24", "24", 0.9),
    SampledPath("(40+8)/2=24", "24", 0.85),
    SampledPath("40/2+4=24", "24", 0.8),
    SampledPath("40-8=32，32/2=16", "16", 0.4),  # 错误
    SampledPath("男生=(40+8)/2=24", "24", 0.75),
]

# 投票
voting = MajorityVoting()
answer, conf = voting.vote(paths)
print(f"问题: {question}")
print(f"答案: {answer} (一致率: {conf:.0%})")
```
