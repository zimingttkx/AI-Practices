## 2. 任务状态机

状态转换演示：

```python
task = Task(name='示例任务')

print('状态转换:')
print(f'初始: {task.status.value}')

task.start()
print(f'开始后: {task.status.value}')

task.complete('完成')
print(f'完成后: {task.status.value}')

task.reset()
print(f'重置后: {task.status.value}')
```

## 3. 层次化分解

创建任务树并分解：

```python
# 创建根任务
root = Task(name='开发Web应用', description='完整的Web应用开发')

# 添加子任务
frontend = Task(name='前端开发', description='React前端')
backend = Task(name='后端开发', description='FastAPI后端')
testing = Task(name='测试', description='单元测试和集成测试')

root.add_subtask(frontend)
root.add_subtask(backend)
root.add_subtask(testing)

print('任务树:')
print(format_task_tree(root))

print(f'\n深度: {root.depth}')
print(f'总子任务数: {root.total_subtasks}')
```

## 4. 依赖分析

拓扑排序演示：

```python
# 创建任务集
tasks = [
    Task(name='需求分析'),
    Task(name='系统设计'),
    Task(name='后端开发'),
    Task(name='前端开发'),
]

# 建立依赖关系
tasks[1].add_dependency(tasks[0].id)  # 设计依赖需求
tasks[2].add_dependency(tasks[1].id)  # 后端依赖设计

# 拓扑排序
analyzer = DependencyAnalyzer()
order = analyzer.topological_sort(tasks)

print('执行顺序:')
for i, task in enumerate(order, 1):
    print(f'{i}. {task.name}')
```

## 5. 分解策略

使用层次化分解器：

```python
decomposer = create_decomposer('hierarchical')

task = Task(
    name='构建电商平台',
    description='完整的电商平台，包括用户管理、商品管理、订单系统'
)

# 分解并附加
result = decomposer.decompose_and_attach(task)

print('分解结果:')
print(format_task_tree(task))
```
